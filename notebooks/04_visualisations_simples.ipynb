Pour faire une bonne visualisation, il faut se poser une question, dans notre cas : Quel semble être le type de tir le plus dangereux (le plus haut pourcentage de buts, pas le plus grand nombre de buts)? Le type de tir le plus courant? 

Pour répondre à cette question, il nous faut deux valeurs, le plus haut pourcentage de buts par type de tirs et la quantité de ce type de tir. 

Pour que ce soit visuellement facile de répondre à ces questions, nous allons faire un diagramme à barres groupés en ordre décroissant sur les tirs totaux. L'axe des x sera les types de tirs, puis l'axe des y sera le nombre de tirs. Il y aura deux barres par type de tirs, l'une sera le nombre de tirs totaux l'autre sera le nombre de buts. Au dessus de la barre des nombres de buts il y aura le pourçentage que cela représente.

```python
season = 20232024
dataframe_season = dataframe[dataframe['season'] == season].copy()
# print(dataframe_season.head(10))

# On va ignorer les tirs sans types
dataframe_season = dataframe_season.dropna(subset=['shot_type'])

df_total_shots_goals = dataframe_season.groupby('shot_type').agg(
    total_shots=('shot_type', 'size'), 
    total_goals=('is_goal', 'sum'))

print(df_total_shots_goals.head(10))
```
