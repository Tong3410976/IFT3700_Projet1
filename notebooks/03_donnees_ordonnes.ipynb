Voici un exemple sur un seul match : 

```python
import json 
from pathlib import Path
import pandas as pd

from ift6758.features.build_dataset import build_dataset
from ift6758.features.shot_features import extract_shots

# Prenons le premier fichier 
path_test = next(Path("../data/raw/").rglob("*.json"))
print(f"Match test : {path_test.name}")

with path_test.open(encoding='utf-8') as f:
    test_json = json.load(f)

data_shots_normalized = pd.read_csv("../data/processed/shots_normalized.csv")

dataframe_test = extract_shots(test_json, data_shots_normalized)
dataframe_test.head()
```
